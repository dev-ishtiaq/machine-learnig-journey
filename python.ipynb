- If/Else

```python
nm = 100
if nm != 10:
    print('true')
else:
    print('false') 
```

```python
md = 10
if md % 2 == 0:
    print('even')
else:
    print('odd')    
```

```python
a,b,c = 17,26,23

if a > b:
    if a > c:
        print('a is greater')
    else: 
        print('c is greater')    

elif b > c:
    if b > a:
        print('b is greater')
    else: 
        print('a is greater') 
else:
    print('c is greater')                   
```

```python
a,b,c = 8,26,23

if a > b and a > c:
    print('a is greater')
elif b > a and b > c:
    print('b is greater')
else:
    print('c is greater')      
```

```python
def fun(name,age):

    print("hello")
    print(name, age)

fun('Ishtiaq',28)    
```

```python
arr = [1,2,3,4,5,6]
def func(array):
    newarr = []
    for num in arr:
        if num % 2 == 0:
            newarr.append(num)

    return newarr

even= func(arr)     
print(even)  
```
