Heat Map আর Pivot টেবল

```python
corr = data.corr(numeric_only=True)
```

```python
sns.heatmap(corr, annot=True)
```

```python
flights = sns.load_dataset('flights')
flights.head()
```

```python
pivot = flights.pivot_table(
    index='month',
    columns= 'year',
    values='passengers'
)
pivot
```

```python
sns.heatmap(pivot, cmap='coolwarm', linewidths=1, linecolor='white')
```

Grid প্লট আর Seaborn এর সমাপ্তি

```python
sns.pairplot(data=data)
```

```python
grid = sns.PairGrid(
    data=data,
    hue='sex'

)
grid.map_diag(sns.histplot)
grid.map_lower(sns.kdeplot, fill=True)
grid.map_upper(sns.kdeplot, fill=True, )
```

```python
sns.set_style('darkgrid')
grid = sns.FacetGrid(
    data=data,
    row='time',
    col='smoker'
)

grid.map(sns.scatterplot,
         'total_bill', 'tip')
```

```python
plt.figure(figsize=(5,5))
sns.countplot(data=data, x= 'day')
```
