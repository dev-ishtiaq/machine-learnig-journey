- MULTI INDEX

```python
groups = ['group1']*3 + ['group2']*3
groups
```

```python
inside_groups = ['row1','row2','row3']*2
inside_groups
```

```python
columns = ['col1','col2','col3']
```

```python
indtup = list(zip(groups, inside_groups))
```

```python
index = pd.MultiIndex.from_tuples(tuples=indtup)
index
```

```python
df4 = pd.DataFrame(
    data = np.random.randn(6,3), 
    index = index, 
    columns = columns
)

df4
```

# missing value handle

```python
data = {
    'A' : [1,2, 3, 4, 5],
    'B' : [5, np.nan, np.nan, 8, 9],
    'C' : [10, 11, 12, np.nan, 14]
}
df = pd.DataFrame(data = data)
df
```

```python
df.dropna(axis=0, thresh=3)
```

```python
df
```

```python
df.fillna(value = 'missing')
```

```python
df.fillna(value = df['A'].mean())
```

```python
df.fillna({'B' : df['B'].mean()})
```

# Pandas group

```python
data2 = {
    'Store Name' : ['Pizza Hut', 'Dominos', 'Pizza Hut', 'Little Caesars', 'Marcos Pizza'],
    'Location' : ['New York', 'Los Angeles', 'Ponchobeki', 'Houston', 'Phoenix'],
    'Price' : np.random.randint(300,500, size=5)
}

df5 = pd.DataFrame(data2)
df5
```

```python
df5.groupby('Store Name').max()
```

```python
store = df5.groupby('Store Name')
store
```

```python
store.mean(numeric_only=True)
```

```python
store.sum(numeric_only=True)
```

```python
store.describe()
```
