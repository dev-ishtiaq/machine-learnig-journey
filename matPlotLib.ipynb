```python
from matplotlib.axes import Axes
fig = plt.figure()
axes : Axes = fig.add_axes([0,0,1,1])
# axes.plot(x,y, color="#02937dff", alpha=0.5, label='X vs Y')
axes.plot(x,x+13, color="#059e24ff", lw=1, ls='-', marker='o', markersize=8, label='X vs X+1')  
axes.plot(x,x+14, color="#f86624ff", lw=1, ls=':', marker='>', markersize=8, label='X vs X+2')
axes.plot(x,x+15, color="#0066ffff", lw=1, ls='-.', marker='<', markersize=8, label='X vs X+3')
axes.plot(x,x+16, color="#bb00dcff", lw=1, ls='-', marker='s', markersize=15,  label='X vs X+4', 
          markerfacecolor='yellow',
            markeredgewidth=3, markeredgecolor='green'
          )
axes.legend(loc='upper left')
```

# Hist

```python
plt.style.use('_mpl-gallery')

# make data
np.random.seed(1)
x = 4 + np.random.normal(0, 1.5, 200)

# plot:
fig, ax = plt.subplots()

ax.hist(x, bins=8, linewidth=0.5, edgecolor="white")

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 56), yticks=np.linspace(0, 56, 9))

plt.show()
```

```python
plt.style.use('_mpl-gallery')

# make the data
np.random.seed(3)
x = 4 + np.random.normal(0, 2, 24)
y = 4 + np.random.normal(0, 2, len(x))
# size and color:
sizes = np.random.uniform(15, 80, len(x))
colors = np.random.uniform(15, 80, len(x))

# plot
fig, ax = plt.subplots()

ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
